```python
from pymongo import MongoClient
import pandas as pd
```

```python
# connect to MongoDB
client = MongoClient('mongodb://localhost:27017')
```

```python
db = client['us_housing_db']
```

```python
collection = db['us_housing']
```

```python
records = collection.find().limit(4)
for record in records:
    print(record,'\n')
```

```python
# find a property based on address
result = collection.find_one({"Address": '208 Michael Ferry Apt. 674\nLaurabury, NE 37010-5101'})
print(result)
```

```python
# find property based on price less than $840,000
query = {"Price": {"$gt": 840000}}

results = collection.find(query)


df = pd.DataFrame(list(results))
print(df.head())
print(df.shape)
```

```python
# find property based on area population less than 3,000
query = {"Area Population": {"$gt": 3000}}

results = collection.find(query)


df = pd.DataFrame(list(results))
print(df.head)
```

```python
most_expensive = collection.find_one(sort=[('Price',-1)])
for index, value in most_expensive.items():
    print(index, value)
```

```python
pipeline = [
    {
        '$group':{
            '_id': None, 
            'average_price': {'$avg': '$Price'}
        }
    }
    
]

result = list(collection.aggregate(pipeline))

average_price = result[0]['average_price']
print(f'Average Price: {average_price}')
```
